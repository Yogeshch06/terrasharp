```python
# Upload exported models to Hugging Face Hub
from huggingface_hub import HfApi, upload_file

api = HfApi()
api.create_repo(repo_id=HF_REPO_ID, token=HF_TOKEN, exist_ok=True)

upload_file(
    path_or_fileobj="terrasharp_swinir_sentinel2.onnx",
    path_in_repo="terrasharp_swinir_sentinel2.onnx",
    repo_id=HF_REPO_ID,
    token=HF_TOKEN
)

upload_file(
    path_or_fileobj="terrasharp_swinir_sentinel2_int8.onnx",
    path_in_repo="terrasharp_swinir_sentinel2_int8.onnx",
    repo_id=HF_REPO_ID,
    token=HF_TOKEN
)

print("Models successfully uploaded to Hugging Face Hub!")
```
